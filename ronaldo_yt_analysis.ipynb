```python
import pandas as pd
import matplotlib.pyplot as plt  # we will use this for plotting
import matplotlib.ticker as ticker
from datetime import datetime, timezone

df = pd.read_csv("cristiano_youtube_stats.csv")

df
```

```python
df.info()
```

```python
df['publishTime'] = pd.to_datetime(df['publishTime']) #Change publishTime to datetime
```

```python
df.head()
```

```python
df_copy = df.copy()
df_copy = df.loc[df_copy.groupby('id')['viewCount'].idxmax()]
df_copy
```

```python
df_copy = df_copy.sort_values(by= 'publishTime', ascending= True)
df_copy
```

```python
df_copy['publishYear'] = df['publishTime'].dt.year
df_copy
```

```python
ax = df_copy[df_copy['video_type'] == 'long'].plot(kind= 'line', x= 'publishTime', y= 'viewCount')
def millions_formatter(x, pos):
    return f'{x*1e-6:.1f}M'
ax.yaxis.set_major_formatter(ticker.FuncFormatter(millions_formatter))

plt.title("Ronaldo's Long Form Videos View Count")
plt.xlabel('Publish Date')
plt.ylabel('Views')
plt.legend().set_visible(False)
plt.show
```

```python
df_copy2 = df.copy()
df_copy2 = df.loc[df_copy2.groupby('id')['viewCount'].idxmax()]
df_copy2 = df_copy2.groupby('video_type')['id'].count()
df_copy2 = df_copy2.reset_index()
df_copy2 = df_copy2.map(lambda x: x.capitalize() if isinstance(x, str) else x)
df_copy2 = df_copy2.set_index('video_type')

df_copy2
```

```python
df_copy2.plot.pie(y= "id", figsize=(6, 6))
plt.ylabel("")
plt.legend().set_title('Content Form')
plt.title('CR7 Youtube Content Distribution')
```

*** Growth Rate of Videos: Use pull_date - publishTime ***

```python
df_copy3 = df.copy()
```

```python
df_copy3['pull_date'] = pd.to_datetime(df_copy3['pull_date'])
df_copy3['publishTime'] = df_copy3['publishTime'].dt.tz_localize(None)
df_copy3['deltaTime'] = df_copy3['pull_date'] - df_copy3['publishTime']
df_copy3['deltaTime'] = df_copy3['deltaTime'].dt.days
df_copy3
```

```python

df_copy3.loc[df_copy3['deltaTime'] < 0, 'deltaTime'] = 0
df_copy3 = df_copy3[df_copy3['deltaTime'] == 60]
df_copy3 = df_copy3.loc[df_copy3.groupby('id')['viewCount'].idxmax()] 
df_copy3 = df_copy3.sort_values('publishTime', ascending=True)
df_copy3 = df_copy3[df_copy3['viewCount'] < 10000000]
df_copy3
```

```python
import matplotlib.dates as mdates

fig, ax = plt.subplots(figsize=(12, 5))
ax.bar(df_copy3['publishTime'], df_copy3['viewCount'], width=3)

ax.xaxis.set_major_locator(mdates.MonthLocator())
ax.xaxis.set_major_formatter(mdates.DateFormatter('%b %Y'))
plt.setp(ax.get_xticklabels(), rotation=45, ha='right')

ax.yaxis.set_major_formatter(ticker.FuncFormatter(millions_formatter))
ax.set_ylim(0, 10_000_000)
ax.set_title('CR7 Videos Views Rate (60 days) without outliers')
ax.set_xlabel('Publish Date')
ax.set_ylabel('Views')
plt.tight_layout()
plt.show()
```
